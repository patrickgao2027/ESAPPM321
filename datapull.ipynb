```python
import pandas as pd
import numpy as np

df = pd.read_csv('Samples_Taxa_ppFEV1_Longitudinal.csv')

taxa_cols = [c for c in df.columns if c not in ['Unnamed: 0', 'ppfev1']]
df['total'] = df[taxa_cols].sum(axis=1)
df['S_rel'] = df['Staphylococcus'] / df['total']
df['P_rel'] = df['Pseudomonas'] / df['total']

# Samples where SA present but PA absent
S_alone = df[(df['S_rel'] > 0) & (df['P_rel'] == 0)]['S_rel'].median()

# Samples where SA present and PA also present
S_with_P = df[(df['S_rel'] > 0) & (df['P_rel'] > 0)]['S_rel'].median()

# Same for PA
P_alone = df[(df['P_rel'] > 0) & (df['S_rel'] == 0)]['P_rel'].median()
P_with_S = df[(df['P_rel'] > 0) & (df['S_rel'] > 0)]['P_rel'].median()

alpha_sp = S_alone / S_with_P
alpha_ps = P_alone / P_with_S

print(f'S alone: {S_alone:.3f}, S with P: {S_with_P:.3f}')
print(f'P alone: {P_alone:.3f}, P with S: {P_with_S:.3f}')
print(f'alpha_sp: {alpha_sp:.3f}')
print(f'alpha_ps: {alpha_ps:.3f}')
```
