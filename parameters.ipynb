```python
import pandas as pd

df = pd.read_csv("C:\\Users\\Owner\\Documents\\Surg docs\\CF_Math\\data\\Samples_Taxa_ppFEV1_Longitudinal.csv")

# Convert sequencing counts to relative abundance
df["SA_rel"] = df["Staphylococcus"] / 2000
df["PA_rel"] = df["Pseudomonas"] / 2000

# Identify species presence
sa_present = df["Staphylococcus"] > 0
pa_present = df["Pseudomonas"] > 0

# Calculate medians among samples where the target species exists
sa_pa_absent = df.loc[
    (~pa_present) & sa_present, "SA_rel"
].median()

sa_pa_present = df.loc[
    pa_present & sa_present, "SA_rel"
].median()

pa_sa_absent = df.loc[
    (~sa_present) & pa_present, "PA_rel"
].median()

pa_sa_present = df.loc[
    sa_present & pa_present, "PA_rel"
].median()

# Calculate exploratory competition coefficients
alpha_sp = sa_pa_absent / sa_pa_present
alpha_ps = pa_sa_absent / pa_sa_present

print("Alpha SP:", alpha_sp)
print("Alpha PS:", alpha_ps)
```
